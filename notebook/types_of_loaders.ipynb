# Creating a sample text file using coding


```python
import os 
os.makedirs("../data/text_files",exist_ok = True)
```

# Sample Text

```python
sample_text = {
    "../data/text_files/python_intro.txt": "Python is a high-level, interpreted programming language known for its simplicity and readability. It supports multiple programming paradigms, including procedural, object-oriented, and functional programming. Python's extensive standard library and active community make it a popular choice for web development, data analysis, artificial intelligence, scientific computing, and more.",
}
for filepath,content in sample_text.items():
    with open(filepath,"w",encoding="utf-8") as f:
        f.write(content)
print("Sample text files created successfully.")
```

# Text Loader


```python

from langchain_community.document_loaders import TextLoader
txt_loader = TextLoader("../data/text_files/python_intro.txt")
documents = txt_loader.load()
print(documents[0].page_content)
```

# Directory Loader

```python
from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader("../data/text_files",
                              glob="**/*.txt",
                              loader_cls=TextLoader,
                              loader_kwargs = {'encoding':'utf-8'})
documents = dir_loader.load()
print(documents[0].page_content)
```
